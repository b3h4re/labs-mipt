```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data =pd.read_excel("data.ods")
data
```

```python
plt.figure(figsize=(12, 6))
plt.grid(True, linestyle="--")
plt.title(r"Зависимость $\beta$ от напряжения")

plt.errorbar(x=data["U, В"], y=data["beta"], fmt="+r", label="Экспериментальные точки")

plt.xlabel(r"$U, В$")
plt.ylabel(r"$\beta$")

plt.legend()
plt.savefig("beta_U_graph.pdf")
plt.show()
```
