Поставить лог лог масштаб на графике

```python
data_vah = pd.read_excel("vah.ods")
data_vah
```

```python
plt.figure(figsize=(12, 6))
plt.grid(True, linestyle="--")
plt.title("ВАХ в двойном логарифмическом масштабе")


plt.xscale('log')
plt.yscale('log')

plt.errorbar(x=data_vah["V"], xerr=0.001, y=data_vah["A"], yerr=0.001, fmt="+r", label="Экспериментальные точки")

plt.xlabel(r"$V, В$")
plt.ylabel(r"$I, А$")


plt.legend()
plt.savefig("log-log.pdf")
plt.show()
```
