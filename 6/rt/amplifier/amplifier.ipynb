```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def print_table_latex(df: pd.DataFrame, order=None, fmt=None):

    if order is not None:
        df_out = df.reindex(columns=order)
    else:
        df_out = df.copy()

    if fmt is not None:
        df_out = df_out.style.format(fmt, na_rep="—")
    
    print(df_out.to_latex())
```

Измерения 2 пункта

```python
U_hat = np.float128('100') # mV
```

```python
data2 = pd.read_csv("data2.csv")
data2
```

```python
data2["K_u"] = data2["A_out"] / data2["A_in"]
data2["K_e"] = data2["A_out"] / U_hat
```

```python
data2
```

```python
fmt = {
    "f": "{:.0f}",
    "A_in": "{:.0f}",
    "A_out": "{:.0f}",
    "K_u": "{:.0f}",
    "K_e": "{:.0f}",
}
print_table_latex(data2, ["f", "A_in", "A_out", "K_u", "K_e"], fmt=fmt)
```

```python
plt.figure(figsize=(12,6))

plt.grid(True, linestyle="--")
plt.title("Зависимость коэффициента усиления от частоты")

plt.errorbar(x=data2["f"], y=data2["K_e"], fmt="+r", label="Экспериментальные точки")

plt.xlabel(r"$f, кГц$")
plt.ylabel(r"$K_e$")

plt.legend()
plt.savefig("pictures/graph_2.pdf")
plt.show()
```

3 пункт

```python
U_hat = np.float128('40') # mV
```

```python
data3 = pd.read_csv("data3.csv")
data3
```

```python
data3["A_out"] = data3["A_out"] * 10
data3["K_u"] = data3["A_out"] / data3["A_in"]
data3["K_e"] = data3["A_out"] / U_hat
```

```python
data3
```

```python
fmt = {
    "f": "{:.0f}",
    "A_in": "{:.0f}",
    "A_out": "{:.0f}",
    "K_u": "{:.0f}",
    "K_e": "{:.0f}",
}
print_table_latex(data3, ["f", "A_in", "A_out", "K_u", "K_e"], fmt=fmt)
```

```python
plt.figure(figsize=(12,6))

plt.grid(True, linestyle="--")
plt.title("Зависимость коэффициента усиления от частоты")

plt.errorbar(x=data3["f"], y=data3["K_e"], fmt="+r", label="Экспериментальные точки")

plt.hlines(max(data3["K_e"]) / np.sqrt(2), xmin=0, xmax=500, colors="black", linestyles="--", label=r"$K_e / \sqrt{2}$")

plt.xlabel(r"$f, кГц$")
plt.ylabel(r"$K_e$")

plt.legend()
plt.savefig("pictures/graph_3.pdf")
plt.show()
```
