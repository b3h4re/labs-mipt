```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Проверка расстояния между питами

```python
CD_distances_data = pd.read_csv("CD/distances.csv")
CD_distances_data.head()
```

```python
R = CD_distances_data["R [nm]"][1:] # Выкинули первую точку, потому что не попадает в 3 сигма

sigma_instrumental = 10*5 # 3 цены деления в нм
sigma_std = np.sqrt(np.std(R, ddof=1))

sigma = np.sqrt(
    sigma_std**2 + sigma_instrumental ** 2
)

print(f"Случайная погрешность измерения: {sigma_std}")
print(f"d = {np.mean(R):.0f} +- {sigma:.0f} нм")
```

Считаем ширину

```python
CD_width_data = pd.read_csv("CD/widths.csv")
CD_width_data.head()
```

```python
w = CD_width_data["R [nm]"] # Выкинули первую точку, потому что не попадает в 3 сигма

sigma_instrumental = 10*5 # 3 цены деления в нм
sigma_std = np.sqrt(np.std(w, ddof=1))

sigma = np.sqrt(
    sigma_std**2 + sigma_instrumental ** 2
)

print(f"Случайная погрешность измерения: {sigma_std:.0f}")
print(f"w = {np.mean(w):.0f} +- {sigma:.0f} нм")
```

Периоды TGT

```python
tgt_periods = pd.read_csv("TGT/periods.csv")
tgt_periods
```

```python
tgt_periods["periods"] = tgt_periods["R [nm]"] / tgt_periods["n_periods"]
tgt_periods
```

```python
sigma_instrumental = 10*5 # 3 цены деления в нм
sigma_std = np.sqrt(np.std(tgt_periods["periods"], ddof=1))

sigma = np.sqrt(
    sigma_std**2 + sigma_instrumental ** 2
)


print(f"T = {np.mean(tgt_periods["periods"]):.0f} +- {sigma:.0f} нм")
```
