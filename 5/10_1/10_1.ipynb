$$ B_\text{мод} = \sqrt2 \cfrac{2 \varepsilon_i}{\pi^2 d^2 N_\text{проб} \nu} $$

$d = (14.9 \pm 0.1)$ мм

```python
e_i = np.float128('7.70e-3') # V
d = np.float128('14.9e-3')   # m

# Частота в розетке
nu = np.float128('50')       # Hz
N_prob = np.float128('44')   # число витков
```

```python
B_mod = 1e3* np.sqrt(2) * 2 * e_i / (np.square(np.pi * d) * N_prob * nu)
print(f"B_mod = {B_mod} мТл")
```

```python
delta_B = (A_1_2 / A_poln) * B_mod
print(f"Delta B = {delta_B}")
```

Калибровка поля и определение g-фактора

```python
# С измерительной катушки
V_0 = np.float128('13.65e-3') # V
```

```python
B_0 = V_0 / (N_prob * (np.pi * d**2 / 4) * (2* np.pi * nu))
print(f"B_0 = {B_0}")
```

$$ g = \cfrac{h f_0}{\mu_B B_0} $$

```python
h_mu_B = np.float128('7.12e-11') # Тл*с
```

```python
g = h_mu_B * f_0 * 1e6 / B_0
print(f"g = {g}")
```
